```python
from tkinter import *
from tkinter import ttk
from PIL import Image, ImageTk
class Personnage:
    def __init__(self, name:str, classe:str, stats:list) -> None:
        self.name = name
        self.classe = classe
        self.cons = stats[0]
        self.stre = stats[1]
        self.agi = stats[2]
        self.inte = stats[3]

    def take_dmg(self, dmg:int):
        self.cons -= dmg
        print(f"{self.name} n'a plus que {self.cons} point de vie")

    def attaque(self, cible):
        dmg = (self.stre*self.agi)//25
        print(f"{self.name} inflige {dmg} points de degats")
        cible.take_dmg(dmg)

    def soin(self):
        print(f"{self.name} se soigne de {self.inte} points")
        self.cons += (self.inte)
        print(f"Il a mtn {self.cons} pv")
def attk():
    Gandalf.attaque(Aragorn)
Aragorn = Personnage("Aragorn", "guerrier", [15, 18, 12, 8])
Gandalf = Personnage("Gandalf", "Mage", [10, 8, 12, 20])
fenetre = Tk()
fenetre.geometry("800x600")
frm=ttk.Frame(fenetre)
frm.grid()
name = ttk.Label(fenetre, text = "Gandalf turn")
img_pil = Image.open("gandalf.png")
img_pil = img_pil.resize((500, 500))
image_tk = ImageTk.PhotoImage(img_pil, master=fenetre)
label_ttk = ttk.Label(fenetre, image=image_tk)
label_ttk.grid(column = 1, row = 1)
name.grid(column = 1, row = 0)
bouton_1 = ttk.Button(fenetre, text = "attaque", command = attk)
bouton_1.grid(column = 0, row = 1)
bouton_2 = ttk.Button(fenetre, text = "soin", command = Gandalf.soin)
bouton_2.grid(column = 2, row = 1)
label_ttk.image = image_tk
fenetre.mainloop()


def main():
    while True:
        print("Tour de Aragorn")
        put=int(input("1 Attaque\n2 Soin\n3 Fuite\n"))
        if put == 1:
            Aragorn.attaque(Gandalf)
        elif put == 2:
            Aragorn.soin()
        elif put ==3:
            break
        print("Tour de Gandalf")
        put=int(input("1 Attaque\n2 Soin\n3 Fuite\n"))
        if put == 1:
            Gandalf.attaque(Aragorn)
        elif put == 2:
            Gandalf.soin()
        elif put ==3:
            break
```
